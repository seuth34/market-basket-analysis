# Exploratory Data Analysis (EDA)

This notebook explores the Instacart Market Basket dataset
before applying the FP-Growth algorithm.

```python
import pandas as pd

orders = pd.read_csv("../data/1_raw/orders.csv")
products = pd.read_csv("../data/1_raw/products.csv")

orders.head()
```

```python
print("Orders Shape:", orders.shape)
print("Products Shape:", products.shape)
```

```python
import matplotlib.pyplot as plt

orders['order_dow'].value_counts().sort_index().plot(kind='bar')

plt.title("Orders by Day of Week")
plt.xlabel("Day")
plt.ylabel("Orders")

plt.show()
```

```python
order_products = pd.read_csv(
    "../data/1_raw/order_products__prior.csv"
)

merged = order_products.merge(
    products,
    on="product_id"
)

top_products = merged['product_name'] \
    .value_counts() \
    .head(10)

print(top_products)
```

```python
top_products.plot(kind='bar')

plt.title("Top 10 Most Purchased Products")
plt.xlabel("Product")
plt.ylabel("Count")

plt.show()
```
